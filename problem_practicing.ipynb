Target element of two sums



```python
def twosum(nums,target):
    d = {}
    for i in range(len(nums)):
        diff = target - nums[i]
        if diff in d:
            return (d[diff],i)
        d[nums[i]] = i
nums = list(map(int, input("Enter numbers: ").split()))
target = int(input("enter target"))
print(twosum(nums,target))
```

 2 duplicate present true


```python
def duplicate(nums):
  for i in range(len(nums)):
    for j in range(i + 1,len(nums)):
      if nums[i] == nums[j]:
        return True
  return False
nums = [1,2,3]
print(duplicate(nums))
```

3 isAnagram or not

```python
def isAnagram(s,t):
  s1 = sorted(s)
  s2 = sorted(t)
  if s1 == s2:
    return True
  return False
s = input("enter s1")
t = input("enter s2")
print(isAnagram(s,t))
```

4 missing number


```python
def missingNumber(nums):
    n = len(nums)

    expected = n * (n + 1) // 2
    actual = sum(nums)

    return expected - actual

nums = list(map(int, input("Enter numbers: ").split()))

print(missingNumber(nums))
```

5 sending zeros to last

```python
def zeroLast(nums):
    j = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[i], nums[j] = nums[j] , nums[i]
            j += 1
    return nums
nums = list(map(int, input("enter values").split()))
print(zeroLast(nums))
```

6 second largest element



```python
def secondLargest(nums):
  nums = list(set(nums))
  nums.sort()
  return nums[-2]
nums = list(map(int,input().split()))
print(secondLargest(nums))
```

palindrome or not


```python
def palindrome(s):
  return s == s[::-1]
s = input("enter a string")
print(palindrome(s))
```

```python
n = int(input())
is_prime = True
for i in range(2, n):
  if n % i == 0:
    is_prime = False
    break
if is_prime:
  print("prime")
else:
  print("not prime")
```
